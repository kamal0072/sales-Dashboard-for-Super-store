```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load datasets
df  = pd.read_csv('Superstore.csv', encoding='latin-1')
df.head(2)
df.tail(2)

df.shape

# for col in df.columns:
#     print(col)

df.dtypes
df.info()
```

```python
# data cleaning
df.isnull().sum()

df[df.duplicated()]

df.drop_duplicates()
df
```

```python
# convert Date
# df['Ship Date'].dtypes
df['Order Date'] = pd.to_datetime(df["Order Date"])
# df['Order Date'].dtype
```

```python
df['Month'] = df['Order Date'].dt.month
df['Year']  = df['Order Date'].dt.year
df.head(3)
```

```python
# # Business KPIs
# # Total Sales, Total Profits, Total Orders, Total Customers , 
total_sales = df['Sales'].sum()
total_profit =  df['Profit'].sum()
total_orders =  df['Order ID'].nunique()
total_customers =  df['Customer ID'].nunique()
# total_customers
# df
```

```python
# Average Order Value
avg_order_value = df['Sales'].mean()
avg_order_value
```

```python
# Now we Perform Exploratory Data Analysis (EDA)
# Monthly Sales Trend
monthly = df.groupby('Month')['Sales'].sum().reset_index()
# monthly
```

```python
# plot
# plt.figure(figsize=(12, 5))
# sns.lineplot(
#     data=monthly,
#     x = 'Month',
#     y=  "Sales",
#     markers="o"
# )
# plt.xticks(rotation = 45)
# plt.show()
```

```python
plt.figure(figsize=(12, 5))
sns.lineplot(
    data=monthly,
    # x=["Jan", 'feb', 'mar', 'apr', 'may', 'june', 'july', 'aug', 'sep', 'oct', 'nov', 'dec'],
    # y=monthly['Sales']
    x="Month",
    y="Sales",
    markers = "o"
)

plt.xlabel("Months")
plt.xticks(rotation = 90)
plt.show()
```

```python
category = df.groupby('Category')['Sales'].sum().reset_index()
category
# sns.barplot(
#     x=category['Category'],
#     y=category['Sales']
# )

```

```python
profit = df.groupby("Category")['Profit'].sum().reset_index()
profit
# sns.barplot(
#     x=profit['Category'],
#     y=profit['Profit']
# )
```

```python
region = df.groupby("Region")["Sales"].sum().reset_index()
region
sns.barplot(
    x=region['Region'],
    y=region['Sales']
)
```

```python
top_10_prducts = df.groupby("Product Name")['Sales'].sum().sort_values(ascending=True)
top_10_prducts
```

```python
# Sales Distribution
sns.histplot(df['Sales'], bins=30, kde=True)
plt.show()
```

```python
# Sales Distribution
sns.boxplot(df['Profit'])
plt.show()
```

```python
# Build a Dashboard Layout
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
# Chart-1
sns.barplot(
    data=category, 
    x="Category",
    y="Sales",
    ax=axes[0, 0]
)
axes[0, 0].set_title("Sales vs. Category", color="red", fontsize=20)
# Chart-2
sns.lineplot(
    data=monthly,
    x="Month",
    y="Sales",
    marker="o",
    ax=axes[0, 1]
)
axes[0, 1].tick_params(axis='x', rotation=45)
axes[0, 1].set_title("Monthly Sales", color="blue", fontsize=20, )
# chart-3
sns.boxplot(
    x='Category',
    y="Profit",
    data=df,
    ax=axes[1, 0]
)
axes[1, 0].set_title("Sales vs Profit", color="pink", fontsize=20)

# Chart-4
sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Sales Vs Profit", color="green", fontsize=20)
plt.tight_layout()
plt.show()
```
